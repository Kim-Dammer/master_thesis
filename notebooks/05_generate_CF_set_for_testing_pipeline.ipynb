## create third setup

coplexes where i have homodimer pdb if the correct stoi is bigger 2, but not necessarily need stoichiometry for other proteins

```python
YM_COMPLEX_overlap = pl.read_csv(
    data_dir / "Complex_Portal/YeastMap/complex_db_complete_unique_overlap_complex_yeastmap_by_complex.csv"
)
complex_db_entries = pl.read_csv(
    data_dir / "Complex_Portal/Saccharomyces_cerevisiae_ComplexTab.tsv",
    separator="	",
)
homodimer_with_pdb = pl.read_csv(
    data_dir / "Pipeline/homodimer_pdb_files/merged_pdbs/pdb_file_names.csv"
)
homodimer_set = set(homodimer_with_pdb["uniprot_id_homodimer"].to_list())

# --- Helpers ---
def _parse_proteins(s):
    if not isinstance(s, str) or not s:
        return []
    return re.findall(r"([A-Za-z0-9_]+)\(\d+\)", s)

def _parse_counts(s):
    if not isinstance(s, str) or not s:
        return []
    return [int(x) for x in re.findall(r"[A-Za-z0-9_]+\((\d+)\)", s)]

def _proteins_with_pdb(s):
    if not isinstance(s, str) or not s:
        return ""
    return ", ".join(p for p in s.split(" ") if p in homodimer_set)

def _pdb_for_true_stoi(stoich_str):
    if not isinstance(stoich_str, str) or not stoich_str:
        return False
    proteins = _parse_proteins(stoich_str)
    counts = _parse_counts(stoich_str)
    if not proteins:
        return False
    return all(
        (cnt < 2) or (prot in homodimer_set)
        for prot, cnt in zip(proteins, counts)
    )

# --- Add stoichiometry info from Complex Portal ---
YM_COMPLEX_overlap = (
    YM_COMPLEX_overlap.join(
        complex_db_entries.select([
            "#Complex ac",
            "Identifiers (and stoichiometry) of molecules in complex",
        ]),
        on="#Complex ac",
        how="left",
    )
    .with_columns(
        stoichiometry_known=(
            pl.col("Identifiers (and stoichiometry) of molecules in complex").is_not_null()
            & ~pl.col("Identifiers (and stoichiometry) of molecules in complex").str.contains(r"\(0\)")
        ),
        solely_proteins=(
            pl.col("Identifiers (and stoichiometry) of molecules in complex").is_not_null()
            & ~pl.col("Identifiers (and stoichiometry) of molecules in complex").str.contains(r":")
        ),
        comb_fold_submission=(
            pl.col("Identifiers (and stoichiometry) of molecules in complex")
            .str.split("|")
            .list.filter(~pl.element().str.starts_with("CHEBI:"))
            .list.filter(~pl.element().str.contains(r"\(0\)"))
            .list.join(",")
        ),
    )
)

# --- Determine CombFold feasibility ---
YM_COMPLEX_overlap = YM_COMPLEX_overlap.with_columns(
    proteins_with_homodimer_pdb=pl.col("true_complex").map_elements(
        _proteins_with_pdb, return_dtype=pl.Utf8,
    ),
    pdb_for_true_stoi=pl.col("comb_fold_submission").map_elements(
        _pdb_for_true_stoi, return_dtype=pl.Boolean,
    ),
)

# --- Filter to complexes where CombFold can run ---
all_pdb_present = YM_COMPLEX_overlap.filter(pl.col("pdb_for_true_stoi"))
```

```python
def make_pairs(proteins: list[str]) -> list[str]:
    proteins = sorted(proteins)
    return [f"{proteins[i]}-{proteins[j]}" for i in range(len(proteins)) for j in range(i+1, len(proteins))]

def explode_pairs(df: pl.DataFrame) -> pl.DataFrame:
    df = df.with_columns(
        pl.col("true_complex").str.split(" ").alias("proteins")
    ).with_columns(
        pl.col("proteins").map_elements(make_pairs, return_dtype=pl.List(pl.String)).alias("pair")
    ).explode("pair").with_columns([
        pl.col("pair").str.split_exact("-", 1).struct.field("field_0").alias("protein_A"),
        pl.col("pair").str.split_exact("-", 1).struct.field("field_1").alias("protein_B"),
    ]).drop("pair", "proteins")

    cols = df.columns
    base = [c for c in cols if c not in ("protein_A", "protein_B")]
    idx = base.index("true_complex") + 1
    return df.select(base[:idx] + ["protein_A", "protein_B"] + base[idx:])

# Usage:
# df = pl.read_csv("no_homodimer_third_setup_results")
# df_long = explode_pairs(df)
```

```python
# Read the iptm parquet
iptm = pl.read_parquet(data_dir/ "iPTM/yeast_pairs_iptm_mapped.parquet")

# Normalize pair ordering in both DataFrames (p1 <= p2) for consistent joins
def join_iptm(pairs_df: pl.DataFrame, iptm_df: pl.DataFrame) -> pl.DataFrame:
    # Normalize iptm pairs
    iptm_norm = iptm_df.with_columns([
        pl.min_horizontal("protein_A", "protein_B").alias("p1"),
        pl.max_horizontal("protein_A", "protein_B").alias("p2"),
    ])

    # Normalize your pairs
    pairs_norm = pairs_df.with_columns([
        pl.min_horizontal("protein_A", "protein_B").alias("p1"),
        pl.max_horizontal("protein_A", "protein_B").alias("p2"),
    ])

    # Left join — keep all your pairs, attach iptm where available
    result = pairs_norm.join(iptm_norm, on=["p1", "p2"], how="left").drop("p1", "p2")

    # Remove duplicate columns from the iptm side (protein_A_right, protein_B_right)
    result = result.drop([c for c in result.columns if c.endswith("_right")])

    return result

# Usage:
no_homodimer_with_iptm = join_iptm(no_homodimer_third_setup_results_pairs, iptm)
```

```python
# === Compare CombFold cluster-0 predictions against biological assembly PDB references ===
# v5: ProDy for RMSD (superpose/calcRMSD) + iterative TM-score (weighted superpose).
#     SIFTS stoichiometry logic preserved. Dead code removed.

import re, csv, json, time, warnings
import urllib.request
from Bio.PDB import PDBParser, MMCIFParser
from scipy.optimize import linear_sum_assignment
from prody import superpose, calcRMSD

warnings.filterwarnings("ignore")
import prody; prody.confProDy(verbosity='none')

# --- Paths ---
combfold_dir = Path(data_dir / "Pipeline/third_setup/CombFold")
assembly_dir = Path(data_dir / "complete_complex_pdb_mapping/_raw_assembly")
assembly_log = Path(data_dir / "complete_complex_pdb_mapping/structure_download_log.csv")

_pdb_parser = PDBParser(QUIET=True)
_cif_parser = MMCIFParser(QUIET=True)

# --- Structure helpers ---

def load_structure(path):
    p = Path(path)
    return _cif_parser.get_structure(p.stem, str(p)) if p.suffix.lower() == ".cif" \
        else _pdb_parser.get_structure(p.stem, str(p))

def chain_ca_map(chain):
    return {r.id[1]: r["CA"].get_coord() for r in chain if r.id[0] == " " and "CA" in r}

def chain_length(chain):
    return sum(1 for r in chain if r.id[0] == " " and "CA" in r)

# --- Geometric core (ProDy) ---

def kabsch_rmsd(P, Q):
    """RMSD after optimal superposition. P=target, Q=mobile."""
    if len(P) == 0: return float("nan")
    Q_aligned, _ = superpose(np.asarray(Q, dtype=float), np.asarray(P, dtype=float))
    return float(calcRMSD(Q_aligned, np.asarray(P, dtype=float)))

def tm_score(P, Q, L_ref):
    """TM-score (Zhang & Skolnick) normalised by L_ref, with iterative
    re-weighting via ProDy's weighted superpose. Converges in ~3-5 iterations."""
    if len(P) == 0 or L_ref == 0: return float("nan")
    d0 = 0.5 if L_ref <= 21 else 1.24 * (L_ref - 15) ** (1/3) - 1.8
    P = np.asarray(P, dtype=float); Q = np.asarray(Q, dtype=float)
    Q_aligned, _ = superpose(Q, P)
    di = np.sqrt(np.sum((Q_aligned - P) ** 2, axis=1))
    for _ in range(5):  # iterative weighting, converges by iter 3
        weights = 1.0 / (1.0 + (di / d0) ** 2)
        Q_aligned, _ = superpose(Q, P, weights=weights)
        di = np.sqrt(np.sum((Q_aligned - P) ** 2, axis=1))
    return float(np.sum(1.0 / (1.0 + (di / d0) ** 2)) / L_ref)

# --- Assembly file discovery ---

def load_assembly_log(log_path):
    if not log_path.exists():
        print(f"  [WARN] Assembly log not found: {log_path}"); return {}
    return {r["pdb_id"].upper(): {"assembly_id": r["assembly_id"], "path": r["path"], "status": r["status"]}
            for r in csv.DictReader(open(log_path, newline="")) if r["pdb_id"]}

def find_assembly_file(pdb_id, asm_log, asm_dir):
    pdb_id = pdb_id.upper()
    entry = asm_log.get(pdb_id)
    if entry and entry["status"] in ("ok", "skip") and entry["path"]:
        p = Path(entry["path"])
        if p.exists(): return (p, entry["assembly_id"], "cif" if p.suffix.lower() == ".cif" else "pdb")
    for pat in [f"{pdb_id}_asm*.pdb", f"{pdb_id}_asm*.cif", f"{pdb_id}_asu.pdb", f"{pdb_id}_asu.cif"]:
        matches = sorted(asm_dir.glob(pat))
        if matches:
            p = matches[0]
            asm_id = p.stem.split("_asm")[-1] if "_asm" in p.stem else "0"
            return (p, asm_id, "cif" if p.suffix.lower() == ".cif" else "pdb")
    return (None, None, None)

# --- SIFTS + UniProt ---

_sifts_cache, _offset_cache, _len_cache = {}, {}, {}

def get_chain_offsets(pdb_id):
    pdb_id = pdb_id.lower()
    if pdb_id in _offset_cache: return _offset_cache[pdb_id]
    offsets = {}
    try:
        with urllib.request.urlopen(f"https://www.ebi.ac.uk/pdbe/api/pdb/entry/residue_listing/{pdb_id}", timeout=30) as r:
            for mol in json.loads(r.read()).get(pdb_id, {}).get("molecules", []):
                for ch in mol.get("chains", []):
                    res = ch.get("residues", [])
                    if res:
                        a, l = res[0].get("author_residue_number"), res[0].get("residue_number")
                        if a is not None and l is not None: offsets[ch.get("struct_asym_id")] = a - l
    except Exception: pass
    _offset_cache[pdb_id] = offsets; time.sleep(0.1); return offsets

def sifts_chain_mapping(pdb_id):
    key = pdb_id.lower()
    if key in _sifts_cache: return _sifts_cache[key]
    rl_offsets = get_chain_offsets(pdb_id)
    chain_map = {}
    try:
        with urllib.request.urlopen(f"https://www.ebi.ac.uk/pdbe/api/mappings/uniprot/{key}", timeout=20) as r:
            data = json.loads(r.read())
            for uniprot, info in data.get(key, {}).get("UniProt", {}).items():
                for m in info.get("mappings", []):
                    ch = m.get("chain_id")
                    if not ch: continue
                    us, ue = m.get("unp_start"), m.get("unp_end")
                    ps, pe = m.get("start", {}).get("residue_number"), m.get("end", {}).get("residue_number")
                    if None in (us, ue, ps, pe): continue
                    author_start = m.get("start", {}).get("author_residue_number")
                    offset = author_start - ps if author_start is not None else rl_offsets.get(m.get("struct_asym_id"), 0)
                    p2u = {ps + i + offset: us + i for i in range(min(ue - us + 1, pe - ps + 1))}
                    if ch in chain_map:
                        chain_map[ch]["pdb_to_uniprot"].update(p2u)
                        chain_map[ch]["unp_end"] = max(chain_map[ch].get("unp_end", 0), ue)
                    else:
                        chain_map[ch] = {"uniprot": uniprot, "pdb_to_uniprot": p2u, "unp_start": us, "unp_end": ue}
    except Exception: pass
    _sifts_cache[key] = chain_map; time.sleep(0.1); return chain_map

def get_uniprot_length(uniprot):
    if uniprot in _len_cache: return _len_cache[uniprot]
    try:
        with urllib.request.urlopen(f"https://rest.uniprot.org/uniprotkb/{uniprot}.fasta", timeout=15) as r:
            length = len("".join(r.read().decode().split("\n")[1:]))
    except Exception: length = None
    _len_cache[uniprot] = length; time.sleep(0.05); return length

# --- Assembly chain collection ---

def collect_assembly_chains(ref_struct, sifts_map):
    chains = []
    for model in ref_struct.get_models():
        for chain in model.get_chains():
            entry = sifts_map.get(chain.id)
            if entry is None and "-" in chain.id:
                entry = sifts_map.get(chain.id.split("-")[0])
            if entry is None: continue
            ca_map = chain_ca_map(chain)
            if ca_map:
                chains.append({"model_id": model.id, "chain_id": chain.id, "uniprot": entry["uniprot"],
                               "sifts": entry, "ca_map": ca_map, "length": len(ca_map)})
    return chains

def assembly_stoichiometry(ref_chain_list):
    stoic = {}
    for rc in ref_chain_list: stoic[rc["uniprot"]] = stoic.get(rc["uniprot"], 0) + 1
    return stoic

def ref_total_ca_count(ref_struct, sifts_map):
    n = 0
    for model in ref_struct.get_models():
        for chain in model.get_chains():
            if chain.id in sifts_map or ("-" in chain.id and chain.id.split("-")[0] in sifts_map):
                n += chain_length(chain)
    return n

# --- Chain matching + evaluation ---

def evaluate_one_reference(cf_struct, ref_struct, sifts_map, uniprot_lengths):
    ref_chain_list = collect_assembly_chains(ref_struct, sifts_map)
    if not ref_chain_list: return float("nan"), float("nan"), {}, {}
    ref_stoic = assembly_stoichiometry(ref_chain_list)

    cf_chains = [{"id": c.id, "ca": chain_ca_map(c)} for c in cf_struct.get_chains() if chain_ca_map(c)]
    if not cf_chains or not ref_chain_list: return float("nan"), float("nan"), {}, ref_stoic

    # Build RMSD cost matrix: CF chain i vs ref chain j, over UniProt-matched Cα
    n_cf, n_ref = len(cf_chains), len(ref_chain_list)
    BIG = 1e6
    rmsd_mat = np.full((n_cf, n_ref), BIG)
    for i, cf in enumerate(cf_chains):
        for j, ref in enumerate(ref_chain_list):
            p2u = ref["sifts"]["pdb_to_uniprot"]
            ref_u_ca = {p2u[r]: c for r, c in ref["ca_map"].items() if r in p2u}
            common = sorted(set(cf["ca"]) & set(ref_u_ca))
            if common:
                rmsd_mat[i, j] = kabsch_rmsd(np.array([cf["ca"][r] for r in common]),
                                             np.array([ref_u_ca[r] for r in common]))

    # Optimal one-to-one assignment (Hungarian)
    size = max(n_cf, n_ref)
    padded = np.full((size, size), BIG); padded[:n_cf, :n_ref] = rmsd_mat
    row_ind, col_ind = linear_sum_assignment(padded)
    matches = {}
    for ri, ci in zip(row_ind, col_ind):
        if ri < n_cf and ci < n_ref and rmsd_mat[ri, ci] < BIG:
            matches[cf_chains[ri]["id"]] = (ref_chain_list[ci]["uniprot"], ref_chain_list[ci])

    if not matches: return float("nan"), float("nan"), {}, ref_stoic

    per_chain_rmsd, cf_all, ref_all = {}, [], []
    for cf_id, (uniprot, ref_chain) in matches.items():
        cf_ca = chain_ca_map(cf_struct[0][cf_id])
        p2u = ref_chain["sifts"]["pdb_to_uniprot"]
        ref_u_ca = {p2u[r]: c for r, c in ref_chain["ca_map"].items() if r in p2u}
        common = sorted(set(cf_ca) & set(ref_u_ca))
        key = f"{uniprot}/{cf_id}"
        if not common: per_chain_rmsd[key] = float("nan"); continue
        cf_coords, ref_coords = np.array([cf_ca[r] for r in common]), np.array([ref_u_ca[r] for r in common])
        per_chain_rmsd[key] = kabsch_rmsd(cf_coords, ref_coords)
        cf_all.append(cf_coords); ref_all.append(ref_coords)

    if not cf_all: return float("nan"), float("nan"), per_chain_rmsd, ref_stoic

    L_ref = ref_total_ca_count(ref_struct, sifts_map)
    tm = tm_score(np.vstack(cf_all), np.vstack(ref_all), L_ref)
    valid = [v for v in per_chain_rmsd.values() if not np.isnan(v)]
    return tm, float(np.mean(valid)) if valid else float("nan"), per_chain_rmsd, ref_stoic

# --- Empty result helper ---

RESULT_COLS = ["best_tm_score", "best_tm_pdb", "best_rmsd", "best_rmsd_pdb",
               "per_chain_rmsd_best", "same_rmsd_and_tm", "discarded_pdb",
               "best_tm_assembly_id", "best_tm_assembly_format", "comparison_status",
               "ref_complete", "cf_stoichiometry", "ref_stoichiometry",
               "missing_uniprots", "extra_ref_uniprots", "coverage_ratio", "n_complete_candidates"]

def _empty(status):
    return dict(zip(RESULT_COLS, [None, "", None, "", "", "", "", "", "", status,
                                  None, "", "", "", "", None, None]))

# --- Main loop ---

print("Loading assembly download log...")
asm_log = load_assembly_log(assembly_log)
print(f"  {len(asm_log)} entries in log.")

results = []
n_scored = n_no_combfold = n_no_ref = n_error = 0
n_total = third_setup_results.height
_t0 = time.time()

for row in third_setup_results.iter_rows(named=True):
    cpx = row["#Complex ac"]
    has_exact = row.get("has_exact_match") == "yes"
    exact_pdbs_str = row.get("exact_match_pdb_ids") or ""

    if len(results) % 10 == 0 and len(results) > 0:
        elapsed = time.time() - _t0
        rate = len(results) / elapsed if elapsed > 0 else 0
        print(f"  [{len(results)}/{n_total}] {cpx} | {elapsed:.0f}s | ~{(n_total - len(results))/rate:.0f}s left", flush=True)

    spec = row.get("comb_fold_submission") or ""
    if not spec:
        results.append(_empty("no_spec")); n_error += 1; continue

    spec_counts = {}
    for m in re.finditer(r"([A-Za-z0-9_]+)\((\d+)\)", spec):
        spec_counts[m.group(1)] = int(m.group(2))
    spec_uniprots = list(dict.fromkeys(spec_counts))
    spec_folder = "_".join(f"{u}x{spec_counts[u]}" for u in sorted(spec_counts))
    cf_path = combfold_dir / f"{spec_folder}_output" / "assembled_results" / "output_clustered_0.pdb"

    try: cf_exists = cf_path.exists()
    except OSError: cf_exists = False
    if not cf_exists:
        results.append(_empty("no_combfold")); n_no_combfold += 1; continue

    if not has_exact or not exact_pdbs_str:
        results.append(_empty("no_reference")); n_no_ref += 1; continue

    try: cf_struct = load_structure(cf_path)
    except Exception:
        results.append(_empty("combfold_load_error")); n_error += 1; continue

    uniprot_lengths = {u: get_uniprot_length(u) for u in spec_uniprots}
    exact_pdbs = [p.strip().upper() for p in exact_pdbs_str.split(";") if p.strip()]

    per_pdb = {}
    for pdb_id in exact_pdbs:
        ref_path, asm_id, asm_fmt = find_assembly_file(pdb_id, asm_log, assembly_dir)
        if ref_path is None: continue
        try: ref_struct = load_structure(ref_path)
        except Exception: continue
        ref_chains = sifts_chain_mapping(pdb_id)
        if not ref_chains: continue
        tm, mean_rmsd, per_chain_rmsd, ref_stoic = evaluate_one_reference(cf_struct, ref_struct, ref_chains, uniprot_lengths)
        if np.isnan(tm): continue
        missing = {u: spec_counts[u] - ref_stoic.get(u, 0) for u in spec_counts if spec_counts[u] > ref_stoic.get(u, 0)}
        extra = {u: ref_stoic[u] - spec_counts.get(u, 0) for u in ref_stoic if ref_stoic[u] > spec_counts.get(u, 0)}
        total_cf = sum(spec_counts.values())
        per_pdb[pdb_id] = {
            "tm": tm, "mean_rmsd": mean_rmsd, "per_chain_rmsd": per_chain_rmsd,
            "asm_id": asm_id, "asm_fmt": asm_fmt, "ref_stoic": ref_stoic,
            "ref_complete": not missing and not extra, "missing": missing, "extra": extra,
            "coverage": sum(min(spec_counts[u], ref_stoic.get(u, 0)) for u in spec_counts) / total_cf if total_cf else 0.0,
        }

    if not per_pdb:
        results.append(_empty("no_overlap")); n_error += 1; continue

    best_tm_pdb = max(per_pdb, key=lambda p: (per_pdb[p]["ref_complete"], per_pdb[p]["tm"]))
    best = per_pdb[best_tm_pdb]
    valid_rmsd = {p: v["mean_rmsd"] for p, v in per_pdb.items() if not np.isnan(v["mean_rmsd"])}
    best_rmsd_pdb = min(valid_rmsd, key=valid_rmsd.get) if valid_rmsd else best_tm_pdb

    pc_str = ";".join(f"{u}:{v:.2f}" for u, v in sorted(best["per_chain_rmsd"].items()) if not np.isnan(v))
    results.append({
        "best_tm_score": round(best["tm"], 4), "best_tm_pdb": best_tm_pdb,
        "best_rmsd": round(best["mean_rmsd"], 2), "best_rmsd_pdb": best_tm_pdb,
        "per_chain_rmsd_best": pc_str,
        "same_rmsd_and_tm": "yes" if best_rmsd_pdb == best_tm_pdb else "no",
        "discarded_pdb": "" if best_rmsd_pdb == best_tm_pdb else best_rmsd_pdb,
        "best_tm_assembly_id": best["asm_id"], "best_tm_assembly_format": best["asm_fmt"],
        "comparison_status": "scored", "ref_complete": best["ref_complete"],
        "cf_stoichiometry": ";".join(f"{u}:{spec_counts[u]}" for u in sorted(spec_counts)),
        "ref_stoichiometry": ";".join(f"{u}:{best['ref_stoic'][u]}" for u in sorted(best["ref_stoic"])),
        "missing_uniprots": ";".join(f"{u}:{best['missing'][u]}" for u in sorted(best["missing"])),
        "extra_ref_uniprots": ";".join(f"{u}:{best['extra'][u]}" for u in sorted(best["extra"])),
        "coverage_ratio": round(best["coverage"], 4),
        "n_complete_candidates": sum(1 for v in per_pdb.values() if v["ref_complete"]),
    })
    n_scored += 1

# Attach results
score_df = pl.DataFrame(results)
existing = [c for c in RESULT_COLS if c in third_setup_results.columns]
if existing: third_setup_results = third_setup_results.drop(existing)
third_setup_results = pl.concat([third_setup_results, score_df], how="horizontal")

# --- Summary ---
print(f"=== Comparison complete ===")
print(f"  Scored: {n_scored} | No CombFold: {n_no_combfold} | No ref: {n_no_ref} | Errors: {n_error}")
scored = third_setup_results.filter(pl.col("comparison_status") == "scored")
n_mismatch = scored.filter(pl.col("same_rmsd_and_tm") == "no").height
print(f"  TM+RMSD same PDB: {scored.height - n_mismatch}/{scored.height} | diff: {n_mismatch}")
print(f"  Format: PDB={scored.filter(pl.col('best_tm_assembly_format')=='pdb').height}, "
      f"mmCIF={scored.filter(pl.col('best_tm_assembly_format')=='cif').height}")
n_complete = scored.filter(pl.col("ref_complete") == True).height
n_incomplete = scored.filter(pl.col("ref_complete") == False).height
print(f"  Stoichiometry complete: {n_complete}/{scored.height} | incomplete: {n_incomplete}/{scored.height}")
if n_incomplete:
    print(f"  avg coverage when incomplete: {scored.filter(pl.col('ref_complete')==False)['coverage_ratio'].mean():.2f}")

print(f"\nTop 10 by TM-score:")
print(scored.sort("best_tm_score", descending=True).select([
    "#Complex ac", "best_tm_score", "best_tm_pdb", "best_rmsd",
    "best_tm_assembly_id", "same_rmsd_and_tm", "discarded_pdb"]).head(10))

mismatch = scored.sort("best_tm_score", descending=True).filter(pl.col("same_rmsd_and_tm") == "no")
print(f"\nTM-best != RMSD-best: {mismatch.height} complexes")
```

```python
# === Visualize CombFold vs reference superposition in Jupyter (FIXED) ===
# Requires: uv pip install py3Dmol
#
# Fixes vs previous version:
#   1. Matches chains by LENGTH (not by order) — same logic as the comparison cell
#   2. Uses Biopython PDBIO (not hand-rolled ATOM lines) — valid PDB output
#   3. Loads reference and transformed CombFold as separate models in py3Dmol

third_setup_results_rmsd = pl.read_csv(data_dir/"Pipeline/third_setup/third_setup_pipeline_complexes_combfold_results_with_exact_match_comparison.csv")
import py3Dmol
import numpy as np
import re
import urllib.request
import json
import time
from pathlib import Path
from Bio.PDB import PDBParser, MMCIFParser, PDBIO, Superimposer
import warnings
warnings.filterwarnings("ignore")

combfold_dir = Path("/cluster/project/beltrao/kdammer/master_thesis/data/Pipeline/third_setup/CombFold")
raw_dir      = Path("/cluster/project/beltrao/kdammer/master_thesis/data/all_Complex_pdb_files/_raw")

_pdb_parser = PDBParser(QUIET=True)
_cif_parser = MMCIFParser(QUIET=True)

def load_structure(path):
    p = Path(path)
    if p.suffix.lower() == ".cif":
        return _cif_parser.get_structure(p.stem, str(p))
    return _pdb_parser.get_structure(p.stem, str(p))

def chain_ca_atoms(chain):
    """List of CA atoms for standard residues in a chain."""
    return [res["CA"] for res in chain if res.id[0] == " " and "CA" in res]

def chain_length(chain):
    return len(chain_ca_atoms(chain))

def get_uniprot_length(uniprot, _cache={}):
    if uniprot in _cache:
        return _cache[uniprot]
    url = f"https://rest.uniprot.org/uniprotkb/{uniprot}.fasta"
    try:
        with urllib.request.urlopen(url, timeout=15) as r:
            fasta = r.read().decode()
        _cache[uniprot] = len("".join(fasta.split("\n")[1:]))
    except Exception:
        _cache[uniprot] = None
    time.sleep(0.05)
    return _cache[uniprot]

def sifts_chain_mapping(pdb_id, _cache={}):
    pdb_id = pdb_id.lower()
    if pdb_id in _cache:
        return _cache[pdb_id]
    url = f"https://www.ebi.ac.uk/pdbe/api/mappings/uniprot/{pdb_id}"
    chain_map = {}
    try:
        with urllib.request.urlopen(url, timeout=20) as r:
            data = json.loads(r.read())
        for uniprot, info in data.get(pdb_id, {}).get("UniProt", {}).items():
            for m in info.get("mappings", []):
                ch = m.get("chain_id")
                if not ch: continue
                us, ue = m.get("unp_start"), m.get("unp_end")
                ps = m.get("start", {}).get("residue_number")
                pe = m.get("end", {}).get("residue_number")
                if None in (us, ue, ps, pe): continue
                p2u = {ps + i: us + i for i in range(min(ue - us + 1, pe - ps + 1))}
                if ch in chain_map:
                    chain_map[ch]["pdb_to_uniprot"].update(p2u)
                else:
                    chain_map[ch] = {"uniprot": uniprot, "pdb_to_uniprot": p2u}
    except Exception:
        pass
    _cache[pdb_id] = chain_map
    time.sleep(0.1)
    return chain_map

def superpose_combfold_on_reference(cf_path, ref_path, spec_uniprots, out_path):
    """Superpose CombFold onto reference by length-matched chains. Writes transformed CombFold PDB."""
    cf = load_structure(cf_path)
    ref = load_structure(ref_path)

    # Get UniProt lengths for the complex's proteins
    uniprot_lengths = {u: get_uniprot_length(u) for u in spec_uniprots}

    # Get reference chain -> UniProt mapping from SIFTS
    ref_pdb_id = ref_path.stem
    ref_chains = sifts_chain_mapping(ref_pdb_id)
    # Build UniProt -> reference chain ID
    uniprot_to_ref_chain = {}
    for rc, info in ref_chains.items():
        u = info["uniprot"]
        if u not in uniprot_to_ref_chain:
            uniprot_to_ref_chain[u] = rc

    # Match CombFold chains to reference chains by length
    cf_atoms_matched = []
    ref_atoms_matched = []
    for cf_chain in cf.get_chains():
        cf_len = chain_length(cf_chain)
        if cf_len == 0:
            continue
        # Find UniProt with this length
        matched_u = None
        for u, ulen in uniprot_lengths.items():
            if ulen == cf_len and u in uniprot_to_ref_chain:
                matched_u = u
                break
        if matched_u is None:
            continue
        ref_chain_id = uniprot_to_ref_chain[matched_u]
        ref_chain = ref[0][ref_chain_id]

        # Get CA atoms, match by UniProt residue position
        cf_ca = chain_ca_atoms(cf_chain)
        ref_ca = chain_ca_atoms(ref_chain)
        p2u = ref_chains[ref_chain_id]["pdb_to_uniprot"]

        # CombFold: resid = UniProt position. Reference: convert pdb resid -> uniprot
        cf_by_pos = {a.get_parent().get_id()[1]: a for a in cf_ca}
        ref_by_pos = {}
        for a in ref_ca:
            resid = a.get_parent().get_id()[1]
            if resid in p2u:
                ref_by_pos[p2u[resid]] = a

        common = sorted(set(cf_by_pos.keys()) & set(ref_by_pos.keys()))
        for pos in common:
            cf_atoms_matched.append(cf_by_pos[pos])
            ref_atoms_matched.append(ref_by_pos[pos])

    if not cf_atoms_matched:
        raise RuntimeError("No chains matched by length — cannot superpose")

    # Superpose
    sup = Superimposer()
    sup.set_atoms(ref_atoms_matched, cf_atoms_matched)
    sup.apply(list(cf.get_atoms()))

    # Write transformed CombFold using PDBIO (valid format)
    io = PDBIO()
    io.set_structure(cf)
    io.save(str(out_path))

    return sup.rms, len(cf_atoms_matched)

# --- Pick a complex to visualize ---
cpx_to_show = "CPX-1424"  # TM 0.98, RMSD 1.06 — your best hit

row = third_setup_results_rmsd.filter(pl.col("#Complex ac") == cpx_to_show).row(0, named=True)
print(f"Showing {cpx_to_show}: TM={row['best_tm_score']}, RMSD={row['best_rmsd']}, ref={row['best_tm_pdb']}")

spec = row["comb_fold_submission"]
spec_uniprots = []
for m in re.finditer(r"([A-Za-z0-9_]+)\((\d+)\)", spec):
    spec_uniprots.append(m.group(1))
spec_uniprots = list(dict.fromkeys(spec_uniprots))

spec_counts = {}
for m in re.finditer(r"([A-Za-z0-9_]+)\((\d+)\)", spec):
    spec_counts[m.group(1)] = int(m.group(2))
spec_folder = "_".join(f"{u}x{spec_counts[u]}" for u in sorted(spec_counts))
cf_path = combfold_dir / f"{spec_folder}_output" / "assembled_results" / "output_clustered_0.pdb"
ref_path = raw_dir / f"{row['best_tm_pdb']}.pdb"
if not ref_path.exists():
    ref_path = raw_dir / f"{row['best_tm_pdb']}.cif"

print(f"  CombFold: {cf_path}")
print(f"  Reference: {ref_path}")

# Superpose (chain-matched, not naive)
out_pdb = Path("/tmp") / f"{cpx_to_show}_combfold_transformed.pdb"
rms, n_atoms = superpose_combfold_on_reference(cf_path, ref_path, spec_uniprots, out_pdb)
print(f"  Chain-matched superposition RMSD: {rms:.2f} Å over {n_atoms} CA atoms")

# Render with py3Dmol
view = py3Dmol.view(width=900, height=600)
view.addModel(ref_path.read_text(), "pdb")
view.setStyle({"model": 0}, {"cartoon": {"color": "gray"}})  # reference = gray
view.addModel(out_pdb.read_text(), "pdb")
view.setStyle({"model": 1}, {"cartoon": {"color": "blue"}})  # CombFold = blue
view.setBackgroundColor("white")
view.zoomTo()
view.show()
```

```python
# Find the complex-ID column
Complex_Portal = pl.read_csv(data_dir/"Complex_Portal/Saccharomyces_cerevisiae_ComplexTab.tsv", separator="\t")
cpx_col = "#Complex ac"

# --- 2. Parse Complex Portal TSV for identity/subset/experimental_evidence tags ---
tsv = pl.read_csv(data_dir/"Complex_Portal/Saccharomyces_cerevisiae_ComplexTab.tsv", separator="	")
WWPDB_TAGGED = re.compile(r"^wwpdb:([A-Za-z0-9]{4})\((identity|subset)\)$")
WWPDB_BARE   = re.compile(r"\bwwpdb:([A-Za-z0-9]{4})\b")

cpx_pdb_info = {}  # cpx -> {"tags": set, "pdb_ids": set}
for row in tsv.iter_rows(named=True):
    cpx = str(row["#Complex ac"]).strip()
    if not cpx:
        continue
    tags = set()
    pdbs = set()
    xref = row.get("Cross references") or ""
    for tok in xref.split("|"):
        m = WWPDB_TAGGED.match(tok.strip())
        if m:
            tags.add(m.group(2))
            pdbs.add(m.group(1).upper())
    expev = row.get("Experimental evidence") or ""
    for match in WWPDB_BARE.finditer(expev):
        pid = match.group(1).upper()
        if pid not in pdbs:
            tags.add("experimental_evidence")
            pdbs.add(pid)
    cpx_pdb_info[cpx] = {"tags": tags, "pdb_ids": pdbs}

# --- 3. Load exact-match CSV ---
exact = pl.read_csv(data_dir/"all_Complex_pdb_files/uniprot_pdb/complex_pdb_exact_match.csv")
exact_info = {}
for row in exact.iter_rows(named=True):
    cpx = row["complex_accession"]
    if row["match_class"] == "exact_match":
        n = int(row["n_exact_match_pdbs"])
        pdbs = row["all_exact_pdbs"] if row["all_exact_pdbs"] is not None else ""
        exact_info[cpx] = {"has_exact": True, "n_exact": n, "pdbs": pdbs}
    else:
        exact_info[cpx] = {"has_exact": False, "n_exact": 0, "pdbs": "NaN"}

# --- 4. Build enrichment columns (Python-side, then attach) ---
cpx_list = Complex_Portal[cpx_col].to_list()
rows = []
for cpx in cpx_list:
    cpx = str(cpx).strip()
    info = cpx_pdb_info.get(cpx, {"tags": set(), "pdb_ids": set()})
    ex = exact_info.get(cpx, {"has_exact": False, "n_exact": 0, "pdbs": ""})
    rows.append({
        "has_complex_portal_pdb":   "yes" if info["pdb_ids"] else "no",
        "complex_portal_tags":      ";".join(sorted(info["tags"])) if info["tags"] else "none",
        "complex_portal_pdb_ids":   ";".join(sorted(info["pdb_ids"])),
        "has_exact_match":          "yes" if ex["has_exact"] else "no",
        "n_exact_match_pdbs":       ex["n_exact"],
        "exact_match_pdb_ids":      ex["pdbs"],
    })

enrich = pl.DataFrame(rows)
Complex_Portal = pl.concat([Complex_Portal, enrich], how="horizontal")

print(f"  Total complexes in CombFold CSV: {Complex_Portal.height}")
n_cp = Complex_Portal .filter(pl.col("has_complex_portal_pdb") == "yes").height
n_id = Complex_Portal.filter(pl.col("complex_portal_tags").str.contains("identity")).height
n_sub = Complex_Portal.filter(pl.col("complex_portal_tags").str.contains("subset")).height
n_ee = Complex_Portal.filter(pl.col("complex_portal_tags").str.contains("experimental_evidence")).height
n_exact = Complex_Portal.filter(pl.col("has_exact_match") == "yes").height
n_neither = Complex_Portal.filter((pl.col("has_complex_portal_pdb") == "no") & (pl.col("has_exact_match") == "no")).height
print(f"  Has Complex Portal PDB (any tag): {n_cp}")
print(f"    of which identity:              {n_id}")
print(f"    of which subset:                {n_sub}")
print(f"    of which experimental_evidence: {n_ee}")
print(f"  Has SIFTS exact match:            {n_exact}")
print(f"  Has neither:                      {n_neither}")
```

How much do the CF prediciton differ

```python
from itertools import combinations
from Bio.PDB import PDBParser, Superimposer


combfold_dir = data_dir / "Pipeline" / "third_setup" / "CombFold"
meta_csv = data_dir / "Pipeline/third_setup/pdb_present_for_stoi_gr_two_third_setup_pipeline_complexes.csv"


_parser = PDBParser(QUIET=True)

def _ca_atoms(path):
    s = _parser.get_structure("s", path)
    atoms = {}
    for chain in s.get_chains():
        for res in chain:
            for atom in res:
                if atom.get_id() == "CA":
                    atoms[(chain.id, res.id[1])] = atom
    return atoms

def rmsd_ca(p1, p2):
    a1, a2 = _ca_atoms(p1), _ca_atoms(p2)
    common = sorted(set(a1) & set(a2))
    if not common:
        return None
    sup = Superimposer()
    sup.set_atoms([a1[k] for k in common], [a2[k] for k in common])
    return round(sup.rms, 3)

# --- folder name from submission: "P05453(1),P12385(1)" -> "P05453x1_P12385x1_output" ---
def submission_to_folder(submission):
    return submission.replace("(", "x").replace(")", "").replace(",", "_") + "_output"

# --- main loop ---
df_meta = pl.read_csv(meta_csv)
rows = []

for row in df_meta.iter_rows(named=True):
    complex_ac = row["#Complex ac"]
    submission = row["comb_fold_submission"]

    results_dir = combfold_dir / submission_to_folder(submission) / "assembled_results"
    try:
        pdbs = sorted(results_dir.glob("output_clustered_*.pdb"),
                      key=lambda p: int(p.stem.split("_")[-1]))
    except OSError:
        pdbs = []  # path too long or missing -> no outputs


    rmsds, pairs = [], []
    for (i, pi), (j, pj) in combinations(enumerate(pdbs), 2):
        r = rmsd_ca(str(pi), str(pj))
        if r is not None:
            rmsds.append(r)
            pairs.append(f"{i}-{j}")

    if rmsds:
        k = rmsds.index(max(rmsds))
        max_rmsd, max_pair = rmsds[k], pairs[k]
    else:
        max_rmsd, max_pair = None, None

    rows.append({
        "complex_ac": complex_ac,
        "comb_fold_submission": submission,
        "n_outputs": len(pdbs),
        "rmsd_list": rmsds,
        "max_rmsd": max_rmsd,
        "max_rmsd_pair": max_pair,
    })

third_setup_results_consistency = pl.DataFrame(rows)
third_setup_results_consistency
```

```python
# add information from which Complex from complex portal the pair is

# --- Build mapping: UniProt protein-set -> Complex Portal accession (CPX-xxxx) ---
cp_tsv = pl.read_csv(data_dir / "Complex_Portal/Saccharomyces_cerevisiae_ComplexTab.tsv", separator="	")
idcol = "Identifiers (and stoichiometry) of molecules in complex"

def extract_uniprot(s):
    """Keep only UniProt IDs (P/O/Q + 5 chars), drop stoichiometry, CPX-, EBI- tokens."""
    if s is None:
        return set()
    ids = []
    for token in re.split(r"[|\s]+", s):
        token = re.sub(r"\(\d+\)", "", token).strip()
        if re.match(r"^[PQO][0-9][A-Z0-9]{4}$", token):
            ids.append(token)
    return set(ids)

# Map: sorted UniProt string -> CPX accession
cp_map = {}
for row in cp_tsv.iter_rows(named=True):
    proteins = extract_uniprot(row[idcol])
    if proteins:
        cp_map[" ".join(sorted(proteins))] = row["#Complex ac"]

# --- Add complex_ac to the pairs table ---
def map_complexes_to_ac(joined_str):
    """Map each CP complex in a ' | '-joined string to its CPX accession."""
    if joined_str is None:
        return None
    acs = []
    for c in joined_str.split(" | "):
        proteins = extract_uniprot(c)
        if proteins:
            key = " ".join(sorted(proteins))
            if key in cp_map:
                acs.append(cp_map[key])
    return " | ".join(sorted(set(acs))) if acs else None

CP_YM_complexes_unique_pairs_CPX_AC = CP_YM_complexes_unique_pairs.with_columns(
    complex_ac=pl.col("Complex_portal_complex").map_elements(map_complexes_to_ac, return_dtype=pl.String)
)
```

```python
# --- Build lookup: uniprot -> paralogue (first row wins if duplicated) ---
paralog_map = (
    biomni_para
    .select("uniprot", "paralog_uniprots")
    .unique(subset=["uniprot"], keep="first")
    .rename({"uniprot": "uniprot", "paralog_uniprots": "paralogue"})
)

# --- Add paralogue columns ---
CP_YM_complexes_unique_pairs_CPX_AC = CP_YM_complexes_unique_pairs_CPX_AC.join(paralog_map.rename({"uniprot": "Protein_1", "paralogue": "p1_paralogue"}), on="Protein_1", how="left")
CP_YM_complexes_unique_pairs_CPX_AC = CP_YM_complexes_unique_pairs_CPX_AC.join(paralog_map.rename({"uniprot": "Protein_2", "paralogue": "p2_paralogue"}), on="Protein_2", how="left")

# --- 1. Original rows (nothing changed) ---
original = CP_YM_complexes_unique_pairs_CPX_AC.with_columns(
    Para_exchange=pl.lit(None, dtype=pl.String),
    Replaced_protein=pl.lit(None, dtype=pl.String),   # ADD THIS
).drop("p1_paralogue", "p2_paralogue")

# --- 2. Prot_1 exchanged ---
p1_exchanged = (
 CP_YM_complexes_unique_pairs_CPX_AC.filter(pl.col("p1_paralogue").is_not_null())
 .with_columns(
    Replaced_protein=pl.col("Protein_1"),   
    Protein_1=pl.col("p1_paralogue"),
    Para_exchange=pl.lit("Prot_1"),
 )
 .drop("p1_paralogue", "p2_paralogue")
)

# --- 3. Prot_2 exchanged ---
p2_exchanged = (
 CP_YM_complexes_unique_pairs_CPX_AC.filter(pl.col("p2_paralogue").is_not_null())
 .with_columns(
    Replaced_protein=pl.col("Protein_2"),   #
    Protein_2=pl.col("p2_paralogue"),
    Para_exchange=pl.lit("Prot_2"),
 )
 .drop("p1_paralogue", "p2_paralogue")
)

# --- 4. Both exchanged ---
both_exchanged = (
 CP_YM_complexes_unique_pairs_CPX_AC.filter(pl.col("p1_paralogue").is_not_null() & pl.col("p2_paralogue").is_not_null())
 .with_columns(
    Replaced_protein=pl.col("Protein_1") + " | " + pl.col("Protein_2"),  
    Protein_1=pl.col("p1_paralogue"),
    Protein_2=pl.col("p2_paralogue"),
    Para_exchange=pl.lit("Both"),
 )
 .drop("p1_paralogue", "p2_paralogue")
)
# --- Combine and drop self-pairs ---
final_cols = ["Protein_1", "Protein_2", "Complex_portal_complex", "YM_complex", "source", "complex_ac", "Para_exchange", "Replaced_protein"]

# --- Combine and drop self-pairs ---
CP_YM_complexes_unique_pairs_CPX_AC_paralogue = (
 pl.concat(
    [original.select(final_cols), p1_exchanged.select(final_cols),
     p2_exchanged.select(final_cols), both_exchanged.select(final_cols)],
    how="vertical",
 )
 .filter(pl.col("Protein_1") != pl.col("Protein_2"))
)
```

```python
# dedublify (paralogue swaps that recreated an existing pair, in either order)
CP_YM_complexes_unique_pairs_dedub_paralogue = (
    CP_YM_complexes_unique_pairs_CPX_AC_paralogue
    .with_columns(
        pair_key=pl.concat_str(
            pl.min_horizontal("Protein_1", "Protein_2"),   
            pl.max_horizontal("Protein_1", "Protein_2"),   
            separator="__",
        )
    )
    .sort(by=pl.when(pl.col("Para_exchange").is_null()).then(0).otherwise(1))  
    .unique(subset="pair_key", keep="first")               
    .drop("pair_key")
)
```

```python
#remove pairs that are with non uniport ids / paralogoue swpa where the replacement was two proteins (with | )
def is_uniprot(s):
    if s is None:
        return False
    return bool(re.match(r"^[PQO][0-9][A-Z0-9]{4}$", s))

CP_YM_complexes_unique_pairs_dedub_paralogue_clean = (
    CP_YM_complexes_unique_pairs_dedub_paralogue
    .filter(
        pl.col("Protein_1").map_elements(is_uniprot, return_dtype=pl.Boolean)
        & pl.col("Protein_2").map_elements(is_uniprot, return_dtype=pl.Boolean)
    )
)
```
