```python
import polars as pl
benchmark_structure_similarity = pl.read_parquet("both_parts_benchmark_structure_similarity.parquet")

df = benchmark_structure_similarity.filter(pl.col("n_proteins") > 2)
df = df.with_columns(
    usalign_pdb_path=pl.col("usalign_pred_dir") + "/complex/usalign.pdb",
    pdb_id=pl.col("reference_pdb_path").str.extract(r"/reference_pdb/([^/]+)/", 1),
).sort("CF_confidence", descending=True).group_by("complex_ac").head(1)

export = df.select(["complex_ac", "pdb_id", "reference_pdb_path", "usalign_pdb_path"])
export.write_csv("export_benchmark_pymol.csv")
print(export.height, "complexes to export")
```
